#### **Data Cleaning**

```python
print(df['Item Fat Content'].unique())
```

```python
df['Item Fat Content']=df['Item Fat Content'].str.lower()
```

```python
df['Item Fat Content'] = df['Item Fat Content'].replace({'reg': 'regular','lf': 'low fat'})
```

```python
df['Item Fat Content'].unique()
```

#### **BUSINESS REQUIREMENTS**

#### **KPI's REQUIREMENTS**

```python
# Total Sales
total_sales=df['Sales'].sum()

# Average Sales
avg_sales=df['Sales'].mean()

# Number of Items
no_of_items_sold=df['Sales'].count()

# Average Rating
avg_ratings=df['Rating'].mean()

# Displaying

print(f"Total Sales: ${total_sales:,.1f}")
print(f"Average Sales: ${avg_sales:,.1f}")
print(f"Number of Items: {no_of_items_sold:,.0f}")
print(f"Average Rating: {avg_ratings:,.1f}")
```

#### **CHART REQUIREMENTS**

#### **Total Sales by Fat Content:**


```python
fat_content = df.groupby('Item Fat Content')['Sales'].sum()

plt.pie(fat_content,labels=fat_content.index,autopct='%1.1f%%',startangle=90)
plt.title("Total Sales by Fat Content")
plt.show()
```

#### **Total Sales by Item Type:**

```python
Item_Type = df.groupby('Item Type',as_index=False)['Sales'].sum().sort_values(by='Sales',ascending=False)

sns.set(rc={'figure.figsize':(15,5)})
ax=sns.barplot(data=Item_Type,x='Item Type',y='Sales',hue='Item Type')
plt.title("Total Sales by Item Type")
plt.xlabel("Item Type")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
for bars in ax.containers:
    ax.bar_label(bars)
plt.show() 
```

#### **Fat Content by Outlet for Total Sales:**

```python

fat_outlet = df.groupby(['Outlet Location Type', 'Item Fat Content'])['Sales'].sum().unstack()

fat_outlet.plot(kind='bar',figsize=(10,6))

plt.title("Fat Content by Outlet for Total Sales")
plt.ylabel("Outlet Location Type")
plt.xlabel("Total Sales")
plt.xticks(rotation=0)

plt.legend(title="Item Fat Content")
plt.show()
```

#### **Total Sales by Outlet Establishment:**

```python
Outlet_sales=df.groupby('Outlet Establishment Year',as_index=False)['Sales'].sum().sort_values(by='Outlet Establishment Year')
sns.lineplot(data=Outlet_sales,x='Outlet Establishment Year',y='Sales',marker='o')
plt.ylabel("Sales")
plt.xlabel("Outlet Establishment Year")
plt.title("Total Sales by Outlet Establishment")
plt.show()
```

#### **Sales by Outlet Size:(Bar Chart)**

```python
Outlet_Size=df.groupby('Outlet Size',as_index=False)['Sales'].sum().sort_values(by='Outlet Size')
sns.set(rc={'figure.figsize':(7,5)})
ax=sns.barplot(data=Outlet_Size,x='Outlet Size',y='Sales',hue='Outlet Size')
for bar in ax.containers:
    ax.bar_label(bar)
plt.ylabel("Size")
plt.xlabel("Total Sales")
plt.title("Total Sales by Outlet Size")
plt.show()
```

#### **Sales by Outlet Size:(Pie Chart)**

```python
Outlet_Size = df.groupby('Outlet Size')['Sales'].sum()
plt.pie(Outlet_Size,labels=Outlet_Size.index,autopct='%1.1f%%')
plt.title("Total Sales by Outlet Size")
plt.show()
```

#### **Sales by Outlet Location:**

```python
Outlet_Location=df.groupby('Outlet Location Type',as_index=False)['Sales'].sum().sort_values(by='Outlet Location Type',ascending=False)
sns.set(rc={'figure.figsize':(10,6)})
ax=sns.barplot(data=Outlet_Location,y='Outlet Location Type',x='Sales')
for bar in ax.containers:
    ax.bar_label(bar)
plt.ylabel("Outlier Location Type")
plt.xlabel("Total Sales")
plt.title("Total Sales by Outlet Location")
plt.show()
```
